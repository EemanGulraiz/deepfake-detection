```python
import torch, timm

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

model = timm.create_model('efficientnet_b0', pretrained=False, num_classes=2)
model.load_state_dict(torch.load('deepfake_detector.pth', map_location=device))
model = model.to(device)
model.eval()
print("Model ready!")
```

```python
import gradio as gr
from torchvision import transforms

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.5,0.5,0.5],[0.5,0.5,0.5])
])

classes = ['fake', 'real']

def predict(img):
    t = transform(img).unsqueeze(0).to(device)
    with torch.no_grad():
        probs = torch.softmax(model(t), dim=1)[0]
    verdict = "🚨 FAKE face detected" if probs[0] > probs[1] else "✅ REAL face detected"
    confidence = {classes[i]: float(f"{probs[i]*100:.1f}") for i in range(2)}
    return verdict, confidence

gr.Interface(
    fn=predict,
    inputs=gr.Image(type="pil", label="Upload a face image"),
    outputs=[gr.Text(label="Verdict"), gr.Label(label="Confidence %")],
    title="Deepfake Face Detector",
    description="Upload any face photo — the AI will tell you if it is real or fake."
).launch(share=True)
```
